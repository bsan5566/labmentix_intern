# **Problem Statement**


This study analyzes the **distribution and diversity of bird species** across two distinct ecosystems — **forests** and **grasslands**. Using bird observation data from both habitats, we examine how environmental factors such as **vegetation type, climate, and terrain** influence bird populations and behavior.

**Objectives:**
- Identify **patterns of habitat preference** among bird species.
- Assess **biodiversity levels** in each ecosystem.
- Evaluate the impact of **environmental conditions** on avian communities.

The findings aim to provide actionable insights for **habitat conservation**, **biodiversity management**, and understanding the **potential effects of environmental changes** on bird populations.

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

### Dataset Loading

```python
# Load Dataset
forest_file = "Bird_Monitoring_Data_FOREST.XLSX"
grassland_file = "Bird_Monitoring_Data_GRASSLAND.XLSX"

df_forest = pd.read_excel(forest_file)
df_grassland = pd.read_excel(grassland_file)
```

### Merging Two datasets

```python
# --- Merge Datasets ---
# Adjust the merge column(s) as per your dataset (example: 'Date' and 'Species')
# If they have same columns, we can simply concatenate:
df_combined = pd.concat([df_forest, df_grassland], ignore_index=True)
```

```python
df_combined.shape
```

### Dataset Information

```python
# Dataset Info
df_combined.info()
```

#### Duplicate Values

```python
# Dataset Duplicate Value Count
df_combined.isnull().sum()
```

```python
df_combined.drop_duplicates(inplace=True)
df_combined.duplicated().sum()
```

#### Missing Values/Null Values

```python
# Missing Values/Null Values Count
df_combined.isnull().sum()
```

```python
# Convert Distance text to numbers (midpoint if range)
df_combined['Distance'] = df_combined['Distance'].replace({
    '<= 50 Meters': 50,
    '50 - 100 Meters': 75,
    '100 - 200 Meters': 150
})

# Fill missing Distance with median
df_combined['Distance'].fillna(df_combined['Distance'].median(), inplace=True)
```

```python
# Fill missing 'Sub_Unit_Code' with 0
df_combined['Sub_Unit_Code'].fillna(0, inplace=True)

# Fill missing 'Site_Name' with 'Unknown Site'
df_combined['Site_Name'].fillna('Unknown Site', inplace=True)

# Fill missing 'ID_Method' with most frequent value
id_method_mode = df_combined['ID_Method'].mode()[0]
df_combined['ID_Method'].fillna(id_method_mode, inplace=True)

# Fill missing 'Distance' with median
distance_median = df_combined['Distance'].median()
df_combined['Distance'].fillna(distance_median, inplace=True)

# Fill missing 'Sex' with 'Unknown'
df_combined['Sex'].fillna('Unknown', inplace=True)

# Fill missing 'AcceptedTSN' with 0
df_combined['AcceptedTSN'].fillna(0, inplace=True)

# Fill missing 'NPSTaxonCode' with 'Unknown'
df_combined['NPSTaxonCode'].fillna('Unknown', inplace=True)

# Fill missing 'TaxonCode' with 0
df_combined['TaxonCode'].fillna(0, inplace=True)

# Fill missing 'Previously_Obs' with 0
df_combined['Previously_Obs'].fillna(0, inplace=True)
```

```python
# Verify no missing values remain
df_combined.isnull().sum()
```
